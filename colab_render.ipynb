```python
# @title 🎬 Step 3: Render Video in Cloud & Auto-Download
# @markdown Select which project you want to render:

target_project = "shop_30s_ad" # @param ["shop_30s_ad", "dukandar_ad_v1", "latest", "alex_talking_demo", "samples"]
auto_download = True # @param {type:"boolean"}
show_preview = True # @param {type:"boolean"}

import os
import glob
from IPython.display import HTML, display
from google.colab import files

print(f"🎬 Rendering target: '{target_project}' on Google Cloud VM...")
!python scripts/cloud_render.py --target "$target_project"

# Locate generated video
video_path = f"exports/{target_project}.mp4"
if not os.path.exists(video_path):
    candidates = sorted(glob.glob("exports/*.mp4"), key=os.path.getmtime, reverse=True)
    if candidates:
        video_path = candidates[0]

if os.path.exists(video_path):
    file_size_mb = os.path.getsize(video_path) / (1024 * 1024)
    print(f"\n🎉 VIDEO RENDERED SUCCESSFULLY IN CLOUD!")
    print(f"📹 Path: {video_path} ({file_size_mb:.2f} MB)")

    if show_preview:
        print("\n🎥 Live Preview:")
        from base64 import b64encode
        with open(video_path, "rb") as f:
            mp4_b64 = b64encode(f.read()).decode()
        html_code = f'''
        <div style="display:flex; justify-content:center;">
          <video width="360" height="640" controls autoplay style="border-radius:18px; box-shadow:0 12px 36px rgba(0,0,0,0.6);">
            <source src="data:video/mp4;base64,{mp4_b64}" type="video/mp4">
          </video>
        </div>
        '''
        display(HTML(html_code))

    if auto_download:
        print("\n📥 Downloading video directly to your computer...")
        files.download(video_path)
else:
    print("❌ Render output not found! Please check output logs above.")
```
