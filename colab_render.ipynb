```python
# @title 🎬 Step 3: Render 30-Second Video & Watch Preview
target_project = "shop_30s_ad" # @param ["shop_30s_ad", "dukandar_ad_v1"]

import os
from IPython.display import HTML, display
from base64 import b64encode

print(f"🎬 Starting render for: '{target_project}' on Google Cloud VM...")
!python scripts/cloud_render.py --target "$target_project"

video_path = f"exports/{target_project}.mp4"
if not os.path.exists(video_path):
    video_path = f"video_output/{target_project}/output/{target_project}.mp4"

if os.path.exists(video_path):
    file_size_mb = os.path.getsize(video_path) / (1024 * 1024)
    print(f"\n🎉 VIDEO RENDERED SUCCESSFULLY IN CLOUD!")
    print(f"📹 Video: {video_path} ({file_size_mb:.2f} MB)")
    print("\n🎥 Playing Live Video below (No auto-download):")
    
    with open(video_path, "rb") as f:
        mp4_b64 = b64encode(f.read()).decode()
        
    html_code = f'''
    <div style="display:flex; flex-direction:column; align-items:center; gap:16px; margin:20px 0;">
      <video width="360" height="640" controls autoplay style="border-radius:18px; box-shadow:0 12px 36px rgba(0,0,0,0.6);">
        <source src="data:video/mp4;base64,{mp4_b64}" type="video/mp4">
        Your browser does not support video playback.
      </video>
      <a href="data:video/mp4;base64,{mp4_b64}" download="{target_project}.mp4" style="background:#10B981; color:#ffffff; padding:12px 28px; border-radius:30px; text-decoration:none; font-weight:800; font-size:16px; box-shadow:0 6px 20px rgba(16,185,129,0.4); display:inline-block;">
        📥 Download {target_project}.mp4 ({file_size_mb:.1f} MB)
      </a>
    </div>
    '''
    display(HTML(html_code))
else:
    print(f"\n❌ Error: Video '{video_path}' was not generated. Please check output logs above.")
```
