```python
# @title 🎬 Step 3: Render Video in Cloud & Watch Preview
# @markdown Select which project you want to render (Default: **apni_dukaan_anime**):

target_project = "apni_dukaan_anime" # @param ["apni_dukaan_anime", "shop_avatar_ad", "shop_30s_ad", "bastar_yash_ad"]
auto_download = False # @param {type:"boolean"}
show_preview = True # @param {type:"boolean"}

import os
import glob
from IPython.display import HTML, display
from google.colab import files

print(f"=======================================================")
print(f"🎯 RENDERING TARGET: '{target_project}'")
if target_project == 'apni_dukaan_anime':
    print(f"📢 Video: 'अपनी दुकान, अपनी पहचान' (LifeSolveNow Anime Ad)")
elif target_project == 'shop_avatar_ad':
    print(f"📢 Video: 'दुकान आपकी, मेहनत आपकी... 0% Commission' (shop LifeSolveNow Talking Avatar Ad)")
elif target_project == 'shop_30s_ad':
    print(f"📢 Video: 'दुकान आपकी, मेहनत आपकी... 0% Commission' (shop LifeSolveNow 30s Ad)")
elif target_project == 'bastar_yash_ad':
    print(f"📢 Video: 'नमस्ते बस्तर! मेरा नाम यश है...' (Full Yash Intro Script)")
print(f"=======================================================\n")

!python scripts/cloud_render.py --target "$target_project"

# Target video path strictly matching the requested project
video_path = f"exports/{target_project}.mp4"
if not os.path.exists(video_path):
    video_path = f"video_output/{target_project}/output/{target_project}.mp4"

if os.path.exists(video_path):
    file_size_mb = os.path.getsize(video_path) / (1024 * 1024)
    print(f"\n🎉 VIDEO RENDERED SUCCESSFULLY IN CLOUD!")
    print(f"📹 Master Video: {video_path} ({file_size_mb:.2f} MB)")

    if show_preview:
        print("\n🎥 Live Preview (Auto-download is OFF - Use the green button to download):")
        from base64 import b64encode
        with open(video_path, "rb") as f:
            mp4_b64 = b64encode(f.read()).decode()
        html_code = f'''
        <div style="display:flex; flex-direction:column; align-items:center; gap:16px; margin:20px 0;">
          <video width="360" height="640" controls autoplay style="border-radius:18px; box-shadow:0 12px 36px rgba(0,0,0,0.6);">
            <source src="data:video/mp4;base64,{mp4_b64}" type="video/mp4">
            Your browser does not support video playback.
          </video>
          <a href="data:video/mp4;base64,{mp4_b64}" download="{target_project}.mp4" style="background:#10B981; color:#ffffff; padding:12px 28px; border-radius:30px; text-decoration:none; font-weight:800; font-size:16px; box-shadow:0 6px 20px rgba(16,185,129,0.4); display:inline-block;">
            📥 Download {target_project}.mp4 ({file_size_mb:.1f} MB)
          </a>
        </div>
        '''
        display(HTML(html_code))

    if auto_download:
        print("\n📥 Auto-downloading video to your computer...")
        files.download(video_path)
else:
    print(f"\n❌ ERROR: Video file '{video_path}' was not generated! Please check the render output logs above.")
```
