## 🎬 Step 3: Assemble Full 3D Cartoon Masterpiece
Raw clips ko hatakar **cinematic animated wipes, slides & iris circle transitions** apply karke theatrical audio mix karte hain:

```python
import subprocess, os
from pathlib import Path

clips_dir = Path("videogen/video_output/shop_digital_dukaan_3d/clips")
clips = [clips_dir / f"clip{i}.mp4" for i in range(1, 7)]
audio_master = "videogen/video_output/shop_digital_dukaan_3d/audio/commercial_audio_master.mp3"
final_commercial = "shop_digital_dukaan_commercial_master.mp4"

# Continuous Broadcast Transitions (xfade):
# 1->2: Smooth Action Slide (slideleft 0.4s at 8.0s)
# 2->3: Storefront Glow Reveal (wipeleft 0.4s at 18.0s)
# 3->4: Dynamic Directional Pan (slideright 0.4s at 26.0s)
# 4->5: Match Cut Delivery Wipe (wipeup 0.4s at 37.0s)
# 5->6: Celebratory Iris-Out Circle Crop (circlecrop 0.4s at 47.0s)

filter_str = (
    "[0:v][1:v]xfade=transition=slideleft:duration=0.4:offset=8.0[v01];"
    "[v01][2:v]xfade=transition=wipeleft:duration=0.4:offset=18.0[v02];"
    "[v02][3:v]xfade=transition=slideright:duration=0.4:offset=26.0[v03];"
    "[v03][4:v]xfade=transition=wipeup:duration=0.4:offset=37.0[v04];"
    "[v04][5:v]xfade=transition=circlecrop:duration=0.4:offset=47.0[vfinal]"
)

cmd = [
    "ffmpeg", "-y",
    "-i", str(clips[0]),
    "-i", str(clips[1]),
    "-i", str(clips[2]),
    "-i", str(clips[3]),
    "-i", str(clips[4]),
    "-i", str(clips[5]),
    "-i", audio_master,
    "-filter_complex", filter_str,
    "-map", "[vfinal]",
    "-map", "6:a",
    "-c:v", "libx264",
    "-preset", "fast",
    "-crf", "19",
    "-c:a", "aac",
    "-b:a", "192k",
    "-shortest",
    "-pix_fmt", "yuv420p",
    final_commercial
]

print("⚡ Rendering Full 3D Pixar Cartoon Commercial in Cloud...")
subprocess.run(cmd, check=True)
print(f"\n🏆 COMMERCIAL MASTER READY! Size: {round(os.path.getsize(final_commercial)/1024/1024, 2)} MB")
```
