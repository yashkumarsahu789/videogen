## 🎬 Step 3: Stitch Seamless TV Cartoon Transitions (Looney Tunes / Cartoon Network Style)
Raw cuts ko hatakar **cinematic animated wipes & circle iris transitions** apply karte hain:

```python
import subprocess, os
from pathlib import Path

clips_dir = Path("videogen/video_output/piku_30s_slapstick/clips")
clips = [clips_dir / f"clip{i}.mp4" for i in range(1, 7)]
audio_master = "videogen/video_output/piku_30s_slapstick/audio/tv_cartoon_audio_master.mp3"
final_tv_output = "piku_tv_cartoon_masterpiece.mp4"

# Continuous Seamless Cartoon Transitions (xfade):
# 1->2: Smooth Action Slide (slideleft 0.35s)
# 2->3: Comic Banana Wipe (wipeleft 0.35s)
# 3->4: Bowling Ball Directional Slide (slideright 0.35s)
# 4->5: Sky Catapult Wipe (wipeup 0.35s)
# 5->6: Classic Looney Tunes Iris-In (circlecrop 0.35s)
filter_str = (
    "[0:v][1:v]xfade=transition=slideleft:duration=0.35:offset=4.65[v01];"
    "[v01][2:v]xfade=transition=wipeleft:duration=0.35:offset=9.30[v02];"
    "[v02][3:v]xfade=transition=slideright:duration=0.35:offset=13.95[v03];"
    "[v03][4:v]xfade=transition=wipeup:duration=0.35:offset=18.60[v04];"
    "[v04][5:v]xfade=transition=circlecrop:duration=0.35:offset=23.25[vfinal]"
)

cmd = [
    "ffmpeg", "-y",
    "-i", str(clips[0]),
    "-i", str(clips[1]),
    "-i", str(clips[2]),
    "-i", str(clips[3]),
    "-i", str(clips[4]),
    "-i", str(clips[5]),
    "-i", audio_master,
    "-filter_complex", filter_str,
    "-map", "[vfinal]",
    "-map", "6:a",
    "-c:v", "libx264",
    "-preset", "fast",
    "-crf", "20",
    "-c:a", "aac",
    "-shortest",
    "-pix_fmt", "yuv420p",
    final_tv_output
]

print("⚡ Rendering Full TV Cartoon Episode in Cloud...")
subprocess.run(cmd, check=True)
print(f"\n🏆 BROADCAST MASTERPIECE READY! Size: {round(os.path.getsize(final_tv_output)/1024/1024, 2)} MB")
```
